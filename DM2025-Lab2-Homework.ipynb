### 1.1 Preprocessing Steps

The text data was cleaned using a lightweight and consistent preprocessing pipeline:

- Normalized and sanitized text (HTML unescape, whitespace cleanup, emoji handling)

- Applied the cleaning function across entire DataFrames - Both *Train* and *Test* data

These steps improved input consistency while preserving the emotional content of the text.

- **Data augmentation** was also tried using an augmentation function; however, this approach **yielded worse results**, likely because the modified texts introduced noise or altered the meaning.

```python
# Create mappings between textual emotion labels and integer class identifiers.
unique_labels = sorted(train_df["emotion"].unique().tolist())
labelId = {label: index for index, label in enumerate(unique_labels)}
idLabel = {index: label for label, index in labelId.items()}

# Convert emotion column to numeric class IDs.
y = train_df["emotion"].map(labelId).values.astype(int)
```

```python
"""
    Getting sample code for LLM (ChatGPT) to generate some data for underrepresented classes.
    This code was used in the initial process for sampling data for sadness, fear, and disgust, to create 
    more data from LLM.
"""

# Map emotions to integers if not already done
train_df['label'] = train_df['emotion'].map({
    'joy': 0, 'anger': 1, 'surprise': 2, 'sadness': 3, 'fear': 4, 'disgust': 5
})

# Sample exactly 30 examples for sadness (3), fear (4), and disgust (5)
sampled_df = train_df.groupby('label').apply(
    lambda x: x.sample(30, random_state=42) if x.name in [1,2,3] else pd.DataFrame()
).reset_index(drop=True)

print("Counts after sampling:")
print(sampled_df['label'].value_counts())

# Map label back to emotion for readability
label2emotion = {0:"joy", 1:"anger", 2:"surprise", 3:"sadness", 4:"fear", 5:"disgust"}
sampled_df['emotion'] = sampled_df['label'].map(label2emotion)

# Select only the id, text, and emotion columns
output_df = sampled_df[['id', 'post', 'emotion']]

# Save to CSV
output_df.to_csv("sampled_emotion_data2.csv", index=False)
print("Saved 90 sampled examples to sampled_emotion_data.csv")

# Preview
print(output_df.head(10))
```

## 3. Model Implementation Steps

```python

# Checking that the torch is supported with cuda, GPU is needed, my laptop is a 4060 
# take around 24 hrs for training, I had to borrow a more powerful GPU machine for training. 
import torch
print(torch.__version__)
print(torch.cuda.is_available()) 
print(torch.cuda.get_device_name(0))  

gpu_available = torch.cuda.is_available()
rocm_available = getattr(torch.version, "hip", None) is not None
fp16_enabled = gpu_available and not rocm_available
bf16_enabled = gpu_available and not rocm_available and torch.cuda.is_bf16_supported()

print(f"rocm available: {rocm_available}")
print(f"Use FP16: {fp16_enabled}")
print(f"Use BF16: {bf16_enabled}")
```

```python
# Setting seeds configs
SEED = 42
set_seed(SEED)
random.seed(SEED)
np.random.seed(SEED)
```

```python
# -----------------------------
# GPU / Precision Setup
# -----------------------------
torch.cuda.empty_cache()

# -----------------------------
# Seeds and model backbones
# -----------------------------
SEED_LIST = [42, 2025]

MODELS = [
    "Amirhossein75/multi-label-emotion-classification-reddit-comments-roberta",
    "cardiffnlp/twitter-roberta-large-emotion-latest",
]

MODEL_PARAMETERS = {
    "Amirhossein75/multi-label-emotion-classification-reddit-comments-roberta": dict(lr=1.5e-5, epochs=4),
    "cardiffnlp/twitter-roberta-large-emotion-latest": dict(lr=1.5e-5, epochs=4),
}

# -----------------------------
# Prepare storage for predictions
# -----------------------------
out_of_fold_predictions_list = []
test_predictions_list = []

# -----------------------------
# Training Loop
# -----------------------------
for model_name in MODELS:
    hp = MODEL_PARAMETERS.get(model_name, {"lr": 2e-5, "epochs": 3})

    for current_seed in SEED_LIST:
        print(f"\nTraining {model_name} - Seed {current_seed}")

        out_of_preds, test_preds = train_transformer_folds(
            model_checkpoint=model_name,
            train_data=train_df,
            test_data=test_df,
            labels_array=y,
            label_to_id=labelId,
            id_to_label=idLabel,
            n_splits=5,
            learning_rate=hp["lr"],
            num_epochs=hp["epochs"],
            train_batch_size=16,
            max_token_length=128,
            weight_decay=0.01,
            seed=current_seed,
            use_fp16=fp16_enabled,
            use_bf16=bf16_enabled,
            checkpoint_dir="./new_trained_model",
        )

        # Accumulate predictions
        out_of_fold_predictions_list.append(out_of_preds)
        test_predictions_list.append(test_preds)

        # Free GPU memory
        torch.cuda.empty_cache()

print("\nTraining completed.")
```

```python
run_all_ensembles(
    out_of_fold_list=out_of_fold_predictions_list,
    test_prediction_list=test_predictions_list,
    ground_truth=y,
    test_df=test_df,
    id_to_label=idLabel
)
```
